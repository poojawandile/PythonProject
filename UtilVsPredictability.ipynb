```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Load the dataset
data = pd.read_csv('C:/Users/HP/SPDelivery.CSV')
data
```

```python
# Create a 2D bar chart for each unique team
unique_teams = data['Team'].unique()
unique_teams
```

```python
# Initialize the plot
fig, axes = plt.subplots(len(unique_teams),1, figsize=(10, 6 * len(unique_teams)), sharex=False)
```

```python
if len(unique_teams) == 1:
    axes = [axes]  # Ensure axes is iterable for a single team case
```

```python
# for team in unique_teams:
#     print(data[data['Team'] == team])
```

```python
#Iterate over each team and plot its data
for i, team in enumerate(unique_teams):
    team_data = data[data['Team'] == team]
    
    # Bar width for differentiation
    bar_width = 0.4
    x_indices = range(len(team_data))
    
    # Plot Utilization and Predictability
    axes[i].bar(x_indices, team_data['Utilization'], width=bar_width, label='Utilization', color='blue')
    axes[i].bar([x + bar_width for x in x_indices], team_data['Predictability'], width=bar_width, label='Predictability', color='orange')
    
    # Set labels and titles
    axes[i].set_title(f"{team} - Utilization and Predictability by Sprint", fontsize=16)
    axes[i].set_xticks([x + bar_width / 2 for x in x_indices])
    axes[i].set_xticklabels(team_data['Sprint_Name'], rotation=45, ha='right', fontsize=12)
    axes[i].set_ylabel("Value (fraction)", fontsize=12)
    axes[i].legend(fontsize=12)
    axes[i].set_ylim(0, 1.2)

# Set common x-axis label
plt.xlabel("Sprint Name", fontsize=14)
plt.tight_layout()
plt.show()
```
